```python
import torch
from torch import nn
```

```python
#create a net to test the various initialisations
net = nn.Sequential(nn.LazyLinear(8), nn.ReLU(), nn.LazyLinear(1))
X = torch.rand(size=(2,4))
net(X).shape
```

```python
#The below example initialises weights nornally and sets bias to zero
def init_normal(module):
    if type(module) == nn.Linear:
        nn.init.normal_(module.weight, mean=0, std=0.01)
        nn.init.zeros_(module.bias)

net.apply(init_normal)
net[0].weight.data.shape, net[0].weight.data[0], net[0].bias.data[0]
```

```python
#you can also initialise layers using provided initialisations like Xavier initialisation
def init_xavier(module):
    if type(module) == nn.Linear:
        nn.init.xavier_uniform_(module.weight)
        nn.init.zeros_(module.bias)

net[0].apply(init_xavier)
print(net[0].weight.data)
```

```python
#parameters can also be accessed and set directly
net[0].weight.data[:] = 1
print(net[0].weight.data)
```
