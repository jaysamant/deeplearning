```python
import torch
from torch import nn 
```

```python
net = nn.Sequential(
    nn.LazyLinear(8),
    nn.ReLU(),
    nn.LazyLinear(1)
)

X = torch.rand(size=(2,4))
net(X).shape
```

```python
#You can access a layer of a model by indexing into the model. You can access the parameters using the indexed layer

#Below code accesses the second linear layer
net[2].state_dict()
```

```python
#Each parameter in pytorch is an instance of Parameter class. Individual values can be accessed using .data attribute
net[2].bias.data
```

```python
#Parameters also contain gradients. The gradient of a parameter can be accessed as follows
net[2].bias.grad
```

```python
#Parameters of all layers can be accessed as follows
[(name, params.shape) for name, params in net.named_parameters()]
```

```python
#You can tie/share model parameters across multiple layers
shared = nn.LazyLinear(8)
net = nn.Sequential(
    nn.LazyLinear(8), nn.ReLU(),
    shared, nn.ReLU(),
    shared, nn.ReLU(),
    nn.LazyLinear(1)
)
net(X)

print(net[2].weight.data[0] == net[4].weight.data[0])
```
