```python
import torch
```

```python
#Scalars and their common operations
x = torch.tensor(3.0)
y = torch.tensor(2.0)
x + y, x * y, x / y, x**y
```

```python
#Vectors are tensors with one dimension
x = torch.arange(4)
x
```

```python
#Elements of a vector can be accessed by indexing
x[0], x[3], x[-1]
```

```python
#The dimensionality of a vector can be accessed by using length or its shape
print(len(x))
print(x.shape) #Shape returns a torch. For a vector, this torch has one element.
```

```python
#Matrices are tensors with two dimensions
A = torch.arange(6).reshape(3, 2)
A
```

```python
#Transpose of a matric is obtained by swapping its axes
A.T
```

```python
#A symmetric matrix is equal to its transpose
A = torch.tensor([[0.0, 1.0], [1.0, 0.0]])
A == A.T
```

```python
#Tensors can have more than two dimensions. # The number of dimensions is called the rank of the tensor. 
# The shape of a tensor is a tuple of integers giving the size of the tensor along each dimension.
X = torch.arange(24).reshape(2, 3, 4)
X
```

```python
#Tensor arithmetric operations apply elementwise
A = torch.arange(6, dtype=torch.float32).reshape(2, 3)
B = A.clone()
A, A + B
```

```python
#Element wise product of two tensors is called the Hadamard product
A * B
```

```python
#Operations between and tensor and scalar apply the operation to each element of the tensor
A, A + 2, A / 2, A**2
```

```python
#Reduction - reduces tensors to scalars. For example, summing all the elements of a tensor reduces it to a scalar.
A, A.sum()
```

```python
#To reduce along a particular axis, we can specify the axis argument. 
# For example, to sum along the rows (axis 0), we can do the following: 
A, A.sum(axis=0)
```

```python
#To reduce along columns (axis 1), we can do the following:
A, A.sum(axis=1)
```

```python
#Mean can ve used to find the average of all elements in a tensor.
A, A.mean(), A.mean(axis=0), A.mean(axis=1)
```

```python
#Non reduction sum keeps the dimensions of the original tensor by setting the keepdims argument to True.
A, A.sum(axis=0, keepdims=True), A.sum(axis=1, keepdims=True)
```

```python
#Dot product of two vectors is the sum of the products of their elements. It will be a scalar.
x = torch.arange(4, dtype=torch.float32)
y = torch.ones(4, dtype=torch.float32) * 2
x, y, torch.dot(x, y)
```

```python
# Dot product between two matrices is the sum of the products of their elements along the last axis of the first matrix and the second to last axis of the second matrix. 
# It will be a matrix.
A = torch.arange(6, dtype=torch.float32).reshape(2, 3)
B = torch.ones(6).reshape(3, 2)
A, B, A@B
```

```python
#Norm is a function that takes a vector as input and outputs a scalar. It is used to measure the magnitude of a vector.
u = torch.tensor([3.0, 4.0])
torch.norm(u) #l2 norm
```

```python
#l1 norm is the sum of the absolute values of the elements of a vector. It is also called the Manhattan norm.
torch.norm(u, p=1)
```

```python
#L2 norm of a matrix is called the Frobenius norm. It is the square root of the sum of the squares of the elements of the matrix.
A = torch.arange(6, dtype=torch.float32).reshape(2, 3)
torch.norm(A)
```
