```python
import torch
```

```python
#Various ways to initialise tensors

x = torch.arange(12,dtype=torch.float32)
x
```

```python
#Determine the number of elements in a tensor
x.numel()
```

```python
#Reshape a tensor
X = x.reshape(3,4)
X
```

```python
#Dimensions of one of the axes can be determined automatically
Y = x.reshape(-1,4)
Y
```

```python
#Initialise a tensor with Zeros
torch.zeros((2,3,4))
```

```python
#Initialise the tensor with ones
torch.ones((2,3,4))
```

```python
#Initialise the tensor with random values drawn from a normal distribution
torch.randn((3,4))
```

```python
#Indexing and slicing tensors
X[-1] #selects the last row of X
```

```python
X[1:3] #selects the 2nd and 3rd row of X
```

```python
#Tensor operations
torch.exp(x)
```

```python
# elementwise operations
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])
x + y, x - y, x * y, x / y, x ** y
```

```python
#Concatenate tensors along rows (axis = 0) or columns (axis = 1)
X = torch.arange(12, dtype=torch.float32).reshape(3,4)
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
torch.cat((X,Y), dim=0), torch.cat((X,Y), dim=1)
```

```python
#Create a binary tensor by comparing two vectors
X == Y
```

```python
#Summing all elements in a tensor
X.sum()
```

```python
#Boardcasting allows us to perform operations on tensors of different shapes. The rules of broadcasting are as follows:
#1. If the tensors do not have the same number of dimensions, prepend the shape of the smaller tensor with 1 until both shapes have the same length.
#2. The two tensors are said to be compatible in a dimension if they have the same size in the dimension, or if one of the tensors has size 1 in that dimension.
#3. The tensors can be broadcast together if they are compatible in all dimensions.

a = torch.arange(3).reshape((3,1))
b = torch.arange(2).reshape((1,2))
a, b
```

```python
a + b
```

```python
#Converting to other python objects
A = X.numpy()
B = torch.tensor(A)
type(A), A, type(B)
```

```python
#converting a tensor to a scalar
a = torch.tensor([3.5])
a, a.item(), float(a), int(a)
```
