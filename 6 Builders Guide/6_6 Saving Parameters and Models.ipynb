```python
import torch
from torch import nn
from torch.nn import functional as F
```

```python
#Tensors can be saved to a file using torch.save
x = torch.arange(4)
torch.save(x, 'x-file')
```

```python
#Tensors can be loaded using torch.load
x2 = torch.load('x-file')
x2
```

```python
#A list of tensors can also be stored to a file like below
y = torch.zeros(4)
torch.save([x, y], 'x-files')

x2, y2 = torch.load('x-files')
(x2, y2)
```

```python
#A dictionary containing tensors can also be written to the disk
mydict = {'x': x, 'y': y}
torch.save(mydict, 'mydict')
mydict2 = torch.load('mydict')
mydict2
```

```python
#We can store the parameters of complete models to the disk
class MLP(nn.Module):
    def __init__(self): 
        super().__init__()
        self.hidden = nn.LazyLinear(256)
        self.out = nn.LazyLinear(10)

    def forward(self, X):
        return self.out(F.relu(self.hidden(X)))

net = MLP()
X = torch.rand(size=(2,20))
Y = net(X)
```

```python
#save the parameters to the disk
torch.save(net.state_dict(), 'mlp.params')
```

```python
#create a model from the parameters stored on the file
clone = MLP()
clone.load_state_dict(torch.load('mlp.params'))
clone.eval()
```
