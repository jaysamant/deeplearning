```python
import torch
from torch import nn
from torch.nn import functional as F
from d2l import torch as d2l
```

```python
#Here we build a layer with no parameters. This layer just centers the data along its mean
class CenteredLayer(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, X):
        return X - X.mean()
```

```python
layer = CenteredLayer()
layer(torch.tensor([[1.0, 2, 3], [4, 5, 6]]))
```

```python
#Now we create a custom layer with parameters.
class MyLinear(nn.Module): 
    def __init__(self, in_units, units):
        super().__init__()
        self.weight = nn.Parameter(torch.randn(in_units, units))
        self.bias = nn.Parameter(torch.randn(units,))

    def forward(self, X):
        linear = torch.matmul(X, self.weight.data) + self.bias.data
        return F.relu(linear)
```

```python
linear = MyLinear(5, 3)
linear.weight
```

```python
linear(torch.rand(2,5))
```
