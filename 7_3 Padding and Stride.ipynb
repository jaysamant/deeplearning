```python
import torch
from torch import nn
```

```python
#Padding - Images are padded with 0 pixels to ensure that pixels at the edges are not ignored
#Without padding after convolution size of image = h-k+1
#With padding p - after comvolution size of image = h-k+1+p. If p=k-1, output size will be same as input size

def comp_conv2d(conv2d, X):
    X = X.reshape((1, 1) + X.shape)
    Y = conv2d(X)
    return Y.reshape(Y.shape[2:])

conv2d = nn.LazyConv2d(1, kernel_size=3, padding=1)
X = torch.rand(size=(8, 8))
comp_conv2d(conv2d, X).shape
```

```python
#Stride is the amount by which you move the kernel. Default is 1
conv2d = nn.LazyConv2d(1, kernel_size=3, padding=1, stride=2)
comp_conv2d(conv2d, X).shape
```
